```python
# Import required libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

print("BOXING FITNESS ANALYZER")
print("Project Started Successfully!")
```

```python
df = pd.read_csv("/content/sample_data/boxing_dataset.csv")
df
```

```python
# Display first 10 records
print("First 10 Records:")
df.head(10)
```

```python
# Display dataset shape
print("\nDataset Shape:")
df.shape
```

```python
# Display column information
print("\nDataset Information:")
df.info()
```

```python
# Check missing and available values

for col in df.columns:
    if col != 'Date':
        available = df[col].notnull().sum()
        missing = df[col].isnull().sum()

        print("\nColumn:", col)
        print("Available Values:", available)
        print("Missing Values:", missing)
```

```python
# Check duplicate workout records

print("\nTotal Duplicate Workout Records:")
print(df.duplicated(subset=['Date', 'Workout_Type', 'Exercise']).sum())
```

```python
# Calculate Training Load

df['Training_Load'] = df['Duration_Min'] * df['RPE']
df
```

```python
# Calculate Total Reps

df['Total_Reps'] = df['Sets'] * df['Reps']
df
```

```python
# Calculate Training Volume

df['Training_Volume'] = df['Sets'] * df['Reps'] * df['Weight_kg']
df
```

```python
# Calculate Running Pace

df['Running_Pace'] = df['Duration_Min'] / df['Distance_km']
df
```

```python
# Calculate Daily Training Summary

daily_summary = df.groupby('Date').agg({
    'Duration_Min': 'sum',
    'Training_Load': 'sum',
    'Exercise': 'count'
})

daily_summary
```

```python
# Calculate Weekly Training Summary

df['Date'] = pd.to_datetime(df['Date'])

df['Week'] = ((df['Date'] - df['Date'].min()).dt.days // 7) + 1

weekly_summary = df.groupby('Week').agg({
    'Duration_Min': 'sum',
    'Training_Load': 'sum',
    'Exercise': 'count'
})

weekly_summary
```

```python
# Calculate Workout Type Summary

workout_summary = df.groupby('Workout_Type').agg({
    'Duration_Min': 'sum',
    'Training_Load': 'sum',
    'Exercise': 'count'
})

workout_summary
```

```python
# Calculate Average Workout Intensity

intensity_summary = df.groupby('Workout_Type')['RPE'].mean()

intensity_summary
```

```python
# Create Fitness Assessment Dataset

fitness_tests = pd.DataFrame({
    'Assessment': ['Initial', 'Final'],
    'Date': [None, None],
    'Cooper_Distance_m': [None, None],
    'Pushups_Reps': [None, None],
    'Broad_Jump_cm': [None, None],
    'Sprint_30m_sec': [None, None],
    'Plank_sec': [None, None]
})

fitness_tests
```

```python
# Cardiovascular Endurance Analysis

running_data = df[
    df['Exercise'].str.contains('run', case=False, na=False)
].copy()

# Calculate daily running performance
cardio_summary = running_data.groupby('Date').agg({
    'Distance_km': 'sum',
    'Duration_Min': 'sum'
})

# Calculate running pace and speed
cardio_summary['Pace_Min_per_km'] = (
    cardio_summary['Duration_Min'] / cardio_summary['Distance_km']
)

cardio_summary['Speed_km_per_hour'] = (
    cardio_summary['Distance_km'] / cardio_summary['Duration_Min'] * 60
)

print("CARDIOVASCULAR ENDURANCE ANALYSIS")
cardio_summary
```
